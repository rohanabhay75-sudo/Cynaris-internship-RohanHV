```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr

data = {
    'Experience': [1, 2, 3, 4, 5],
    'Salary': [20000, 25000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

pearson = pearsonr(df['Experience'], df['Salary'])
spearman = spearmanr(df['Experience'], df['Salary'])

print("Pearson Correlation:", pearson.statistic)
print("Spearman Correlation:", spearman.statistic)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Correlation matrix
correlation = df.corr()

# Heatmap
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# Scatter plot
sns.scatterplot(x='Experience', y='Salary', data=df)
plt.title("Experience vs Salary")
plt.show()

# Scatter matrix
pd.plotting.scatter_matrix(df, figsize=(6, 6))
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

# Prepare data
X = df[['Experience']]
y = df['Salary']

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict salary
predictions = model.predict(X)

print("Predicted Salaries:", predictions)
print("Slope:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

r2 = r2_score(y, predictions)
mae = mean_absolute_error(y, predictions)
rmse = np.sqrt(mean_squared_error(y, predictions))

print("R2 Score:", r2)
print("MAE:", mae)
print("RMSE:", rmse)
```

```python
import matplotlib.pyplot as plt

# Calculate residuals
residuals = y - predictions

# Plot residuals
plt.scatter(predictions, residuals)
plt.axhline(y=0, color='red', linestyle='--')

plt.xlabel("Predicted Salary")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
```
