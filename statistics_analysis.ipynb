```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data= {'Employee':['A','B','C','D','E','F','G','H','I','J'],
'Sales':[12000,15000,18000,20000,22000,25000,28000,30000,35000,60000]}
df=pd.DataFrame(data)
df
```

```python
mean_value=df['Sales'].mean()
print("Mean:",mean_value)
```

```python
median_value=df['Sales'].median()
print("meadian:",median_value)
```

```python
mode_value=df['Sales'].mode()
print("mode:",mode_value)
```

```python
variance_value=df['Sales'].var()
print("Variance:",variance_value)
```

```python
std_value=df['Sales'].std()
print("Standard deviation:",std_value)
```

```python
print("Mean:",mean_value)
print("Median:",median_value)
print("Mode:",mode_value)
print("Variance:",variance_value)
print("Standard Deviation:",std_value)
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Sales'],bins=5,edgecolor='black')
plt.title('Sales Frequency Distribution')
plt.xlabel('Sales Amount')
plt.ylabel('Number of Employees')
plt.show()
```

```python
# Business Observation:

# The histogram shows that most employees have sales
# between ₹12,000 and ₹35,000. One employee has
# significantly higher sales of ₹60,000.

# This indicates a right-skewed distribution.
# The higher sales value may influence the average.
```

```python
print("Mean:", df['Sales'].mean())
print("Median:", df['Sales'].median())

if df['Sales'].mean() > df['Sales'].median():
    print("Distribution: Right Skewed")
elif df['Sales'].mean() < df['Sales'].median():
    print("Distribution: Left Skewed")
else:
    print("Distribution: Approximately Symmetric")
```

```python
skewness = df['Sales'].skew()

print("Skewness:", skewness)
```

```python
np.random.seed(42)

normal_data = np.random.normal(50, 10, 1000)

right_skewed_data = np.random.exponential(10, 1000)

bimodal_data = np.concatenate([
    np.random.normal(30, 5, 500),
    np.random.normal(70, 5, 500)
])

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(normal_data, kde=True, ax=axes[0])
axes[0].set_title("Normal Distribution")

sns.histplot(right_skewed_data, kde=True, ax=axes[1])
axes[1].set_title("Right-Skewed Distribution")

sns.histplot(bimodal_data, kde=True, ax=axes[2])
axes[2].set_title("Bimodal Distribution")

plt.tight_layout()

plt.show()
```

```python
p25 = df['Sales'].quantile(0.25)
p50 = df['Sales'].quantile(0.50)
p75 = df['Sales'].quantile(0.75)
p90 = df['Sales'].quantile(0.90)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)
print("90th Percentile:", p90)
```

```python
Q1 = df['Sales'].quantile(0.25)
Q2 = df['Sales'].quantile(0.50)
Q3 = df['Sales'].quantile(0.75)

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1

print("Interquartile Range:", IQR)
```

```python
group_a = [12000, 15000, 18000, 20000, 22000]

group_b = [25000, 28000, 30000, 35000, 60000]

print("Group A Average:", np.mean(group_a))
print("Group B Average:", np.mean(group_b))
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    [group_a, group_b],
    tick_labels=['Group A', 'Group B']
)

plt.title('Sales Comparison Between Two Employee Groups')
plt.xlabel('Employee Groups')
plt.ylabel('Sales Amount')

plt.show()
```

```python
print("Group A Median:", np.median(group_a))
print("Group B Median:", np.median(group_b))

print("Group A Standard Deviation:", np.std(group_a, ddof=1))
print("Group B Standard Deviation:", np.std(group_b, ddof=1))
```
