# Week 4 - Day 4: Hypothesis Testing

## Objective
To conduct hypothesis tests and validate business claims using statistical evidence.

## Scenario 1: E-commerce Delivery
H0: Average delivery time = 3 days.
H1: Average delivery time != 3 days.

## Scenario 2: Employee Productivity
H0: Average tasks completed = 10.
H1: Average tasks completed != 10.

## Scenario 3: Marketing Campaign
H0: New campaign does not increase average sales.
H1: New campaign increases average sales.

```python
import numpy as np
from scipy import stats
```

```python
delivery_days = np.array([2.5, 3.2, 3.5, 2.8, 3.1,
                          3.6, 2.9, 3.4, 3.0, 3.3])

print("Delivery Days:", delivery_days)
print("Average Delivery Time:", np.mean(delivery_days))
```

```python
t_stat, p_value = stats.ttest_1samp(delivery_days, popmean=3)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject Null Hypothesis (H0)")
    print("Average delivery time is significantly different from 3 days.")
else:
    print("Fail to Reject Null Hypothesis (H0)")
    print("No significant evidence that delivery time differs from 3 days.")
```

```python
import numpy as np
from scipy import stats

company_A = np.array([2.5, 3.0, 2.8, 3.2, 2.9, 3.1, 2.7, 3.0])

company_B = np.array([3.5, 3.8, 3.6, 4.0, 3.7, 3.9, 3.6, 3.8])

print("Company A Average:", np.mean(company_A))
print("Company B Average:", np.mean(company_B))
```

```python
t_stat, p_value = stats.ttest_ind(company_A, company_B)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject Null Hypothesis (H0)")
    print("There is a significant difference between delivery times.")
else:
    print("Fail to Reject Null Hypothesis (H0)")
    print("There is no significant evidence of a difference.")
```

```python
delivery_days = np.array([2.5, 3.2, 3.5, 2.8, 3.1,
                          3.6, 2.9, 3.4, 3.0, 3.3])

confidence_interval = stats.t.interval(
    0.95,
    df=len(delivery_days)-1,
    loc=np.mean(delivery_days),
    scale=stats.sem(delivery_days)
)

print("Mean:", np.mean(delivery_days))
print("95% Confidence Interval:", confidence_interval)
```

## Type I and Type II Errors

### Type I Error
Rejecting a true null hypothesis.
Example: Concluding that an ineffective medicine works.

### Type II Error
Failing to reject a false null hypothesis.
Example: Missing a marketing campaign that actually increases sales.

### Business Importance
These errors help companies understand the risks of making decisions based on statistical tests.

```python
delivery_days = np.array([2.5, 3.2, 3.5, 2.8, 3.1,
                          3.6, 2.9, 3.4, 3.0, 3.3])

t_stat, p_value = stats.ttest_1samp(delivery_days, 3)

print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Delivery time is significantly different from 3 days.")
else:
    print("Fail to reject H0: No significant difference found.")
```

```python
company_A = np.array([2.5, 3.0, 2.8, 3.2, 2.9, 3.1, 2.7, 3.0])

company_B = np.array([3.5, 3.8, 3.6, 4.0, 3.7, 3.9, 3.6, 3.8])

t_stat, p_value = stats.ttest_ind(company_A, company_B)

print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Significant difference between companies.")
else:
    print("Fail to reject H0: No significant difference found.")
```

```python
before_campaign = np.array([100, 110, 105, 115, 108, 112, 103, 107])

after_campaign = np.array([125, 130, 128, 135, 132, 127, 140, 133])

t_stat, p_value = stats.ttest_ind(
    before_campaign,
    after_campaign,
    alternative='less'
)

print("Before Campaign Average:", np.mean(before_campaign))
print("After Campaign Average:", np.mean(after_campaign))
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0: Evidence suggests the campaign increased sales.")
else:
    print("Fail to reject H0: Insufficient evidence of increased sales.")
```
