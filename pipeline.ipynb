```python
import pandas as pd
```

```python
orders=pd.read_csv(r'C:\Users\ROHAN\Downloads\da_b2b_orders.csv', parse_dates=['order_date'])
orders=orders.drop_duplicates("order_id")
```

```python
calc = orders.quantity_units * orders.unit_price_inr
print("Value mismatches:", ((calc - orders.total_value_inr).abs() > 1).sum())
print(orders.isna().sum())

metrics = {
    "total_orders": len(orders),
    "total_revenue_inr": orders.total_value_inr.sum(),
    "avg_order_value_inr": orders.total_value_inr.mean(),
    "on_time_rate": orders.on_time_delivery.mean(),
    "avg_delivery_days": orders.delivery_days.mean(),
    "return_rate": orders.return_qty.sum() / orders.quantity_units.sum(),
    "repeat_buyer_rate": orders.repeat_buyer.mean(),
}
```

```python
pd.Series(metrics).to_csv(r"C:\Users\ROHAN\OneDrive\Desktop\Cynaris_Internship folder\TradeSight Export Ananlysis\outputs\baseline_metrics.csv")

(orders.groupby("category").total_value_inr.sum()
       .sort_values(ascending=False).to_csv("outputs/revenue_by_category.csv"))
(orders.groupby(orders.order_date.dt.to_period("M")).total_value_inr.sum()
       .to_csv("outputs/monthly_revenue.csv"))
```
